### Machine Learning Workflow for Eclipse Classification

To build a model that predicts the **Eclipse Type** (e.g., Total, Annular, Partial, Hybrid), we follow a standard machine learning pipeline:

#### 1. Data Preprocessing & Target Definition
* **Dependent Variable (Target / $Y$):** `Eclipse Type` (categorical label).
* **Independent Variables (Features / $X$):** Features like `Gamma`, `Eclipse Magnitude`, `Sun Altitude`, `Sun Azimuth`, etc.
* *Note:* We must first convert any non-numerical values (like the '-' dashes in `Path Width`) into clean numbers, and encode our target labels into numbers (e.g., Partial = 0, Total = 1, etc.).

#### 2. Train-Test Splitting
To make sure our model actually learns patterns rather than just memorizing the data, we split our dataset into two parts:
* **Training Set (typically 80%):** The data the model uses to learn the relationships between features and the target.
* **Testing Set (typically 20%):** Hidden data used to evaluate how accurately our model can predict unseen future eclipses.

#### 3. Choosing the Model: ID3 vs. Random Forest

* **ID3 (Iterative Dichotomiser 3 / Decision Tree):**
  * This is a fundamental tree-based algorithm. It splits the data step-by-step by calculating **Information Gain** (using the Entropy concept we discussed earlier!).
  * *Pros:* Very easy to visualize and explain (we can plot the exact rule-set of the tree).
  * *Cons:* Can easily "overfit" (become too specific to the training set and perform poorly on new data).

* **Random Forest:**
  * Instead of relying on just one decision tree, a Random Forest builds an "ensemble" (a forest) of many different decision trees trained on random subsets of the data.
  * When making a prediction, every tree gets a "vote", and the majority vote wins.
  * *Pros:* Highly accurate, robust, and prevents overfitting.
  * *Cons:* More complex and harder to visualize than a single tree.

### Step 1: Preprocessing and Target Encoding
We will replace missing placeholder dashes (`-`) with actual numerical values or handle them, cast numeric columns correctly, and encode the `Eclipse Type` column into numeric labels.

```python
import numpy as np

# 1. Clean 'Path Width (km)' by replacing '-' with NaN and converting to float
df['Path Width (km)'] = pd.to_numeric(df['Path Width (km)'].replace('-', np.nan), errors='coerce')

# Since Path Width represents the width of the shadow path, missing/null values are likely
# Partial Eclipses (where there is no central path of totality). We can fill these NaNs with 0.
df['Path Width (km)'] = df['Path Width (km)'].fillna(0)

# 2. Let's see our unique values in the target 'Eclipse Type'
print("Original Eclipse Types:", df['Eclipse Type'].unique())

# We often simplify these types into broader classes (e.g., T=Total, A=Annular, P=Partial, H=Hybrid)
# Let's map them to numerical categories:
# P = 0, A = 1, T = 2, H = 3
type_mapping = {'P': 0, 'A': 1, 'T': 2, 'H': 3}

# Map the target column. If there are sub-types (like 'T-' or 'A+'), we extract the first letter.
df['Eclipse Class'] = df['Eclipse Type'].str[0].map(type_mapping)

# Drop any row where the mapping couldn't be resolved (if any)
df = df.dropna(subset=['Eclipse Class'])
df['Eclipse Class'] = df['Eclipse Class'].astype(int)

print("Encoded target distribution:")
print(df['Eclipse Class'].value_counts())

# Display cleaned dataframe info
df.info()
```
